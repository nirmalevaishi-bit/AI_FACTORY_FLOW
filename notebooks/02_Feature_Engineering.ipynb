# Stage 10 - Feature Engineering

## Project

AI-Based Smart Factory Flow Optimizer

## Objective

Prepare the dataset for Machine Learning by creating new features.

```python
import pandas as pd
import numpy as np
```

```python
df = pd.read_csv("../data/processed/cleaned_factory_flow_data.csv")
```

```python
df.head()
```

```python
df.info()
```

```python
df["Worker_Status"] = df["Worker_Available"].map({
    "Yes":1,
    "No":0
})
```

```python
df[["Worker_Available","Worker_Status"]].head()
```

## Observation

Worker availability has been converted into numerical values.

Yes = 1

No = 0

```python
df["Machine_Load"] = df["Machine_Status"].map({
    "Idle":0,
    "Busy":1,
    "Maintenance":2,
    "Overloaded":3
})
```

```python
df[["Machine_Status","Machine_Load"]].head()
```

```python
df["Shift_Code"] = df["Shift"].map({
    "Morning":0,
    "Afternoon":1,
    "Night":2
})
```

```python
df[["Shift","Shift_Code"]].head()
```

```python
def queue_category(q):

    if q < 5:
        return "Low"

    elif q < 15:
        return "Medium"

    else:
        return "High"

df["Queue_Category"] = df["Queue_Length"].apply(queue_category)
```

```python
df["Queue_Category"].value_counts()
```

```python
def temp_level(t):

    if t < 50:
        return "Normal"

    elif t < 70:
        return "Warm"

    else:
        return "Hot"

df["Temperature_Level"] = df["Temperature"].apply(temp_level)
```

```python
df["Temperature_Level"].value_counts()
```

```python
df["Bottleneck_Flag"] = np.where(
    df["Queue_Length"] >= 15,
    1,
    0
)
```

```python
df["Bottleneck_Flag"].value_counts()
```

```python
def delay_level(wait):

    if wait < 10:
        return "Low"

    elif wait < 20:
        return "Medium"

    else:
        return "High"

df["Delay_Level"] = df["Expected_Waiting_Time"].apply(delay_level)
```

```python
df["Delay_Level"].value_counts()
```

```python
df.head()
```

```python
df.to_csv("../data/processed/final_factory_data.csv", index=False)

print("Feature Engineering Completed Successfully!")
```
